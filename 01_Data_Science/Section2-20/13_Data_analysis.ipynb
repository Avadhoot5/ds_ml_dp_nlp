### 03. Data Manipulation With Pandas

```python
import pandas as pd
```

```python
df = pd.read_csv('contents/employees.csv')
```

```python
df.head()
```

```python
df.tail()
```

```python
df.describe()
```

```python
df.dtypes
```

```python
df.isnull()
```

```python
df.isnull().any()
```

```python
df.isnull().sum()
```

```python
# df_filled = df.fillna(0)
```

```python
# Rename columns

df = df.rename(columns={'Company':'Company Name'})
```

```python
df.info()
```

```python
df['Salary_fill'] = df['Salary'].fillna(df['Salary'].mean())
```

```python
df['New Date'] = pd.to_datetime(df['Start Date'], errors='coerce')
df.head()
```

```python
df['Updated City'] = df['City'].apply(lambda x:f'City {x}')
```

```python
df.head()
```

```python
# Data aggregating and grouping

grouped_mean = df.groupby('Team')['Salary'].mean()
print(grouped_mean)
```

```python
grouped_sum = df.groupby('Team')['Salary'].sum()
print(grouped_sum)
```

```python
df.groupby(['Team', 'Gender'])['Salary'].mean()
```

```python
# multiple aggregate functions

grouped_agg = df.groupby('Team')['Salary'].agg(['mean','sum','count'])
grouped_agg
```

```python
# Merging and joining data frames

df1 = pd.DataFrame({'Key': ['A', 'B', 'C'], 'Value1': [1,2,3]})
df2 = pd.DataFrame({'Key': ['A', 'B', 'D'], 'Value2': [4,5,6]})

print(df1)
print(df2)
```

```python
# Merge df on the 'Key' column

pd.merge(df1,df2, on='Key', how='inner')
```

```python
pd.merge(df1,df2, on='Key', how='outer')
```

```python
pd.merge(df1,df2, on='Key', how='left')
pd.merge(df1,df2, on='Key', how='right')
```
