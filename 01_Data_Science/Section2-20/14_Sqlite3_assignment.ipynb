```python
# ### Assignment 1: Creating and Connecting to a Database

# 1. Write a Python function to create a new SQLite3 database named `test.db`.

def create_db(db_name):
    connection = sqlite3.connect(db_name)
    return connection

connection = create_db('contents/test.db')

# 2. Write a Python function to create a table named `employees` with columns `id` (integer), `name` (text), `age` (integer), and `department` (text) in the `test.db` database.

def create_table(connection, table_name):
    cursor = connection.cursor()
    cursor.execute(f'''
    CREATE TABLE IF NOT EXISTS {table_name}(
        id Integer PRIMARY KEY,
        name Text NOT NULL,
        age Integer,
        department Text
    )
    ''')

create_table(connection,'employees')
connection.commit()

connection.close()
```

```python
# ### Assignment 3: Querying Data

# 1. Write a Python function to fetch and display all records from the `employees` table.

connection = create_db('contents/test.db')

def fetch_all_records(connection):
    cursor = connection.cursor()
    cursor.execute('''
    SELECT * FROM employees    
    ''',)
    rows = cursor.fetchall()
    for row in rows: print(row)

# 2. Write a Python function to fetch and display all employees from a specific department.

def fetch_records_by_dept(connection, department_name):
    cursor = connection.cursor()
    cursor.execute('''
    SELECT * FROM employees
    WHERE department = ?
    ''', (department_name, ))
    rows = cursor.fetchall()
    for row in rows: print(row)

fetch_all_records(connection)
print('-' * 50)
fetch_records_by_dept(connection, 'Finance')
connection.close()
```

```python
# ### Assignment 4: Updating Data

# 1. Write a Python function to update the department of an employee based on their `id`.

connection = create_db('contents/test.db')

def update_dept(connection, emp_id, new_dept):
    cursor = connection.cursor()
    cursor.execute('''
    UPDATE employees
    SET department = ?
    WHERE id = ?
    ''', (new_dept, emp_id))
    connection.commit()
    print('Data Set updated')

# 2. Update the department of at least 2 employees and display the updated records.

def update_depts(connection, data, emp_id):
    cursor = connection.cursor()
    cursor.executemany('''
    UPDATE employees
    SET department = ?
    WHERE id = ?
    ''', (data))
    connection.commit()
    
    print('Data Set updated')

    placeholder = ', '.join('?' for _ in emp_id)

    # print the updated records
    cursor.execute(f'''
    SELECT * FROM employees
    WHERE id IN ({placeholder})
    ''', emp_id)
    rows = cursor.fetchall()
    for row in rows: print(row)

update_dept(connection, 5, 'Finance')

data = [('Finance', 5), ('Data Analyst', 7)]
update_depts(connection, data, (5, 7))

connection.close()
```

```python
# ### Assignment 6: Advanced Queries

# 1. Write a Python function to fetch and display employees older than a certain age.

connection = create_db('contents/test.db')

def fetch_emp_old(connection, older_age):
    cursor = connection.cursor()
    cursor.execute('''
    SELECT * FROM employees
    WHERE age > ?    
    ''', (older_age,))
    rows = cursor.fetchall()
    print(f'Below are the employees older than {older_age}')
    for row in rows: print(row)

# 2. Write a Python function to fetch and display employees whose names start with a specific letter.

def fetch_emp_name_start(connection, letter):
    cursor = connection.cursor()
    cursor.execute('''
    SELECT * FROM employees
    WHERE name LIKE ?
    ''', (letter + '%', ))
    rows = cursor.fetchall()
    print(f'Below are the employees whose names start with {letter}')
    for row in rows: print(row) 

fetch_emp_old(connection, 30)
print('-' * 50)
fetch_emp_name_start(connection, 'A')

connection.close()
```

```python
# ### Assignment 7: Handling Transactions

# 1. Write a Python function to insert multiple employees into the `employees` table in a single transaction. Ensure that if any insertion fails, none of the insertions are committed.

connection = create_db('contents/test.db')

def insert_employees(connection, data):
    cursor = connection.cursor()

    try:
        cursor.executemany('''
        INSERT INTO employees (name, age, department)
        VALUES (?, ? , ?)
        ''', data)

        connection.commit()
        print('All employees inserted successfully!')

    except Exception as e:
        connection.rollback()
        print(f'Insertion failed: {e}')

data = [
    ('Andrew', 30, 'Data Scientist'),
    ('Sarah', 28, 'Software Engineer'),
    ('Michael', 35, 'Data Analyst'),
    ('Emily', 25, 'Web Developer'),
    ('David', 32, 'Project Manager'),
    ('Jessica', 27, 'HR'),
    ('Daniel', 34, 'Finance'),
    ('Olivia', 29, 'Marketing'),
    ('James', 31, 'IT'),
    ('Sophia', 26, 'Sales'),
    ('William', 38, 'Operations'),
    ('Ava', 24, 'Data Analyst'),
    ('Benjamin', 33, 'Cybersecurity'),
    ('Isabella', 28, 'Finance'),
    ('Ethan', 36, 'Software Engineer')
]


# 2. Write a Python function to update the age of multiple employees in a single transaction. Ensure that if any update fails, none of the updates are committed.

def update_age(connection, age_id):
    cursor = connection.cursor()

    try:
        cursor.executemany('''
        UPDATE employees
        SET age = ?
        WHERE id = ?
        ''', age_id)
        connection.commit()
        print('Age updated successfully')
    except Exception as e:
        connection.rollback()
        print(f'Updation failed: {e}')

insert_employees(connection, data)
print('-' * 50)
update_age(connection, [(35, 7), (25, 8)])

connection.close()
```

```python
# ### Assignment 8: Creating Relationships

# 1. Create a new table named `departments` with columns `id` (integer) and `name` (text).

connection = create_db('contents/test.db')

def create_table(connection):
    try:
        cursor = connection.cursor()
        cursor.execute('''
        CREATE TABLE IF NOT EXISTS departments (
            id Integer PRIMARY KEY,
            name Text NOT NULL
        )
        ''')
        print('Table Created sucess')
        connection.commit()
    except Exception as e:
        connection.rollback()
        print(f'Table creation failed: {e}')

create_table(connection)

# 2. Modify the `employees` table to include a foreign key referencing the `id` column in the `departments` table.

def modify_employees(connection):
    try:
        cursor = connection.cursor()
        cursor.execute('''
        CREATE TABLE IF NOT EXISTS employees_new (
            id INTEGER PRIMARY KEY,
            name TEXT NOT NULL,
            age INTEGER,
            department_id INTEGER,
            FOREIGN KEY (department_id) REFERENCES departments(id)
        )
        ''')
        print('Modified the employees table')
        connection.commit()
    except Exception as e:
        connection.rollback()
        print(f'Table modification failed: {e}')

# 3. Write a Python function to insert data into both the `departments` and `employees` tables, ensuring referential integrity.

def insert_data(connection, dept_data, emp_data):
    try:
        cursor = connection.cursor()
        
        cursor.executemany('''
            INSERT INTO departments (id, name)
            VALUES (?, ?)
        ''', dept_data)

        cursor.executemany('''
            INSERT INTO employees_new
            VALUES (?, ?, ?, ?)
        ''', emp_data)

        connection.commit()
        print(f'Inserted data in both tables successfully')

    except Exception as e:
        connection.rollback()
        print(f'Table Insertion failed: {e}')

create_table(connection)
modify_employees(connection)

dept_data = [(1, 'Finance')]
emp_data = [(1,'Andrew', 35, 1)]
insert_data(connection, dept_data, emp_data)

connection.close()
```
