```python
### Assignment 11: Abstract Base Class
# Create an abstract base class named `Shape` with an abstract method `area`. Create derived classes `Circle` and `Square` that implement the `area` method. Create objects of the derived classes and call the `area` method.

from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return f'The area of the circle is {3.14 * self.radius * self.radius}'

class Square(Shape):
    def __init__(self, size):
        self.size = size

    def area(self):
        return f'The area of the Square is {self.size * self.size}'

circle1 = Circle(5)
print(circle1.area())

square1 = Square(5)
print(square1.area())
```

```python
### Assignment 14: Class with Context Manager
# Create a class named `FileManager` that implements the context manager protocol to open and close a file. Use this class to read the contents of a file.

class FileManager:
    def __init__(self, filename):
        self.filename = filename

    def __enter__(self):
        self.file = open(f'contents/{self.filename}', 'r')
        return self.file

    def __exit__(self, exc_type, exc, tb):
        self.file.close()

with FileManager('data.txt') as file:
    contents = file.read()
    print(contents)
```

```python
### Assignment 5: Encapsulation with Private Attributes
# Create a class named `BankAccount` with private attributes `account_number` and `balance`. Add methods to deposit and withdraw money, and to check the balance. Ensure that the balance cannot be accessed directly.

class BankAccount:
    def __init__(self, account_number, balance):
        self.__account_number = account_number
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def deposit(self, amount):
        self.__balance += amount


    def withdraw(self, money):
        try:
            if money > self.__balance:
                raise Exception('Cannot withdraw money greater than the balance')
            else:
                self.__balance -= money
        except Exception as e:
            print(e)

bank = BankAccount(101, 5000)

print(bank.get_balance())
bank.deposit(2000)
print(bank.get_balance())
bank.withdraw(20000)
print(bank.get_balance())
```

```python
### Assignment 6: Encapsulation with Property Decorators
# In the `BankAccount` class, use property decorators to get and set the `balance` attribute. Ensure that the balance cannot be set to a negative value.

class BankAccount:
    def __init__(self, account_number, balance):
        if balance < 0:
            print('Please provide a valid amount')
        self.__account_number = account_number
        self.__balance = balance

    @property
    def balance(self):
        return self.__balance

    @balance.setter
    def balance(self, money):
        if money < 0:
            print('Please provide a valid amount')
        else:
            self.__balance = money

bank = BankAccount(101, 5000)

print(bank.balance)
bank.balance = 2000
print(bank.balance)
```

```python
### Assignment 9: Abstract Methods in Base Class
# Create an abstract base class named `Employee` with an abstract method `calculate_salary`. Create two derived classes `FullTimeEmployee` and `PartTimeEmployee` that implement the `calculate_salary` method. Create objects of the derived classes and call the `calculate_salary` method.

from abc import ABC, abstractmethod

class Employee(ABC):
    @abstractmethod
    def calculate_salary(self):
        pass

class FullTimeEmployee(Employee):
    def __init__(self, salary):
        self.salary = salary

    def calculate_salary(self):
        return self.salary * 4

class PartTimeEmployee(Employee):
    def __init__(self, salary):
        self.salary = salary

    def calculate_salary(self):
        return self.salary * 2

fte = FullTimeEmployee(50000)
pte = PartTimeEmployee(30000)

print(fte.calculate_salary())
print(pte.calculate_salary())
```

```python
### Assignment 10: Encapsulation in Data Classes
# Create a data class named `Product` with private attributes `product_id`, `name`, and `price`. Add methods to get and set these attributes. Ensure that the price cannot be set to a negative value.

class Product:
    def __init__(self, product_id, name, price):
        self.__product_id = product_id
        self.__name = name
        if (price < 0): print('Price cannot be a negative value')
        else: self.__price = price

    @property
    def product_id(self):
        return self.__product_id

    @product_id.setter
    def product_id(self, new_id):
        self.__product_id = new_id
    
    @property
    def name(self):
        return self.__name

    @name.setter
    def name(self, new_name):
        self.__name = new_name
    
    @property
    def price(self):
        return self.__price

    @price.setter
    def price(self, new_price):
        if (new_price < 0): print('Price cannot be a negative value')
        else: self.__price = new_price

phone = Product(101, 'SmartPhone', 30000)

print(phone.product_id)
print(phone.name)
print(phone.price)

phone.product_id = 105
phone.name = 'Tablet'
phone.price = 45000

print(phone.product_id)
print(phone.name)
print(phone.price)
```

```python
### Assignment 11: Polymorphism with Operator Overloading
# Create a class named `Vector` with attributes `x` and `y`. Overload the `+` operator to add two `Vector` objects. Create objects of the class and test the operator overloading.

class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

vector1 = Vector(5, 2)
vector2 = Vector(2, 8)

result = vector1 + vector2
print(result.x, result.y)
```

```python
### Assignment 12: Abstract Properties
# Create an abstract base class named `Appliance` with an abstract property `power`. Create two derived classes `WashingMachine` and `Refrigerator` that implement the `power` property. Create objects of the derived classes and access the `power` property.

from abc import ABC, abstractmethod

class Appliance(ABC):
    def __init__(self, power):
        self._power = power

    @property
    @abstractmethod
    def power(self):
        pass

class WashingMachine(Appliance):
    def __init__(self, power):
        super().__init__(power)

    @property
    def power(self):
        return (f'The Washing machine can {self._power}')

    @power.setter
    def power(self, new_power):
        self._power = new_power

class Refrigerator(Appliance):
    def __init__(self, power):
        super().__init__(power)

    @property
    def power(self):
        return (f'The Refrigerator machine can {self._power}')

    @power.setter
    def power(self, new_power):
        self._power = new_power

wm = WashingMachine('wash clothes')
rf = Refrigerator('cool things')

print(wm.power)
print(rf.power)
```

```python
### Assignment 13: Encapsulation in Class Hierarchies
# Create a base class named `Account` with private attributes `account_number` and `balance`. Add methods to get and set these attributes. Create a derived class named `SavingsAccount` that adds an attribute `interest_rate`. Create an object of the `SavingsAccount` class and test the encapsulation.

class Account:
    def __init__(self, account_number, balance):
        self.__account_number = account_number
        if (balance < 0): print('Balance cannot be negative')
        else:
            self.__balance = balance

    @property
    def account_number(self):
        return self.__account_number

    @account_number.setter
    def account_number(self, new_account):
        self.__account_number = new_account

    @property
    def balance(self):
        return self.__balance

    @balance.setter
    def balance(self, new_balance):
        if (new_balance < 0): print('Provide proper balance amount')
        else:
            self.__balance = new_balance

class SavingsAccount(Account):
    def __init__(self, account_number, balance, interest_rate):
        super().__init__(account_number, balance)
        self.interest_rate = interest_rate


person = SavingsAccount(105, 2000, 0.5)

print(person.account_number)
print(person.balance)
print(person.interest_rate)

person.account_number = 110
person.balance = 10000

print(person.account_number)
print(person.balance)
print(person.interest_rate)
```

```python
### Assignment 14: Polymorphism with Multiple Inheritance
# Create a class named `Flyer` with a method `fly`. Create a class named `Swimmer` with a method `swim`. Create a class named `Superhero` that inherits from both `Flyer` and `Swimmer` and overrides both methods. Create an object of the `Superhero` class and call both methods.
 
class Flyer:
    def fly(self):
        print('Flyer is flying')

class Swimmer:
    def swim(self):
        print('Swimmer is swimming')

class Superhero(Flyer, Swimmer):

    def fly(self):
        print('Superhero is flying')

    def swim(self):
        print('Superhero is swimming')

sh = Superhero()
sh.fly()
sh.swim()
```
