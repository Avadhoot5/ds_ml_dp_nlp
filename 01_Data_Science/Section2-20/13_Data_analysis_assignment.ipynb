```python
# ### Assignment 1: Array Creation and Manipulation

# 1. Create a NumPy array of shape (5, 5) filled with random integers between 1 and 20. Replace all the elements in the third column with 1.
# 2. Create a NumPy array of shape (4, 4) with values from 1 to 16. Replace the diagonal elements with 0.

arr1 = np.array([[1 if (j==2) else random.randint(1,20) for j in range(5)] for i in range(5)])

print(arr1)
print(arr1.shape)

print('=' * 20)

arr2 = np.arange(1, 17).reshape(4, 4)
np.fill_diagonal(arr2, 0)

print(arr2)
print(arr2.shape)
```

```python
# ### Assignment 2: Array Indexing and Slicing

# 1. Create a NumPy array of shape (6, 6) with values from 1 to 36. Extract the sub-array consisting of the 3rd to 5th rows and 2nd to 4th columns.
# 2. Create a NumPy array of shape (5, 5) with random integers. Extract the elements on the border.

arr1 = np.arange(1, 37).reshape(6, 6)

print(arr1)

sub_arr1 = arr1[2:5, 1:4]

print('='*15)

print(sub_arr1)

print('='*15)

arr2 = np.array([[np.random.randint(25) for _ in range(5)] for _ in range(5)])

print(arr2)

zero_arr = np.zeros(arr2.shape, dtype=bool)

zero_arr[0,:] = True
zero_arr[-1,:] = True
zero_arr[:,0] = True
zero_arr[:,-1] = True

border_elements = arr2[zero_arr]

print('='*15)

print(border_elements)
```

```python
# ### Assignment 3: Array Operations

# 1. Create two NumPy arrays of shape (3, 4) filled with random integers. Perform element-wise addition, subtraction, multiplication, and division.
# 2. Create a NumPy array of shape (4, 4) with values from 1 to 16. Compute the row-wise and column-wise sum.

arr1 = np.random.randint(1, 20, size=(3,4))
arr2 = np.random.randint(1, 20, size=(3,4))

print(f'Addition \n {arr1 + arr2}')
print(f'Subtraction \n {arr1 - arr2}')
print(f'Multiplication \n {arr1 * arr2}')
print(f'Division \n {arr1 / arr2}')

arr3 = np.arange(1, 17).reshape(4, 4)

row_sum = np.sum(arr3, axis=1)
col_sum = np.sum(arr3, axis=0)

print(f'Row sum: {row_sum}')
print(f'Col sum: {col_sum}')
```

```python
# ### Assignment 4: Statistical Operations

# 1. Create a NumPy array of shape (5, 5) filled with random integers. Compute the mean, median, standard deviation, and variance of the array.
# 2. Create a NumPy array of shape (3, 3) with values from 1 to 9. Normalize the array (i.e., scale the values to have a mean of 0 and a standard deviation of 1).

arr1 = np.random.randint(1, 20, size=(5,5))
print(f'Mean: {arr1.mean()}')
print(f'Median: {np.median(arr1)}')
print(f'SD: {arr1.std()}')
print(f'Variance: {arr1.var()}')

print("-" * 40)

arr2 = np.arange(1, 10).reshape(3, 3).astype(float)

mean = np.mean(arr2)
std_dev = np.std(arr2)

normalized_matrix = (arr2 - mean) / std_dev

print(f'Mean of normalized matrix: {np.mean(normalized_matrix):.1f}')
print(f'STD of normalized matrix: {np.std(normalized_matrix):.1f}')
```

```python
# ### Assignment 5: Broadcasting

# 1. Create a NumPy array of shape (3, 3) filled with random integers. Add a 1D array of shape (3,) to each row of the 2D array using broadcasting.
# 2. Create a NumPy array of shape (4, 4) filled with random integers. Subtract a 1D array of shape (4,) from each column of the 2D array using broadcasting.

arr1 = np.random.randint(1, 10, size=(3,3))
row_to_add  = np.array([10, 20, 30])

print(f'Orig Arr: \n {arr1}')
print("-" * 40)
result = arr1 + row_to_add
print(f'Result array: \n {result}')
print("-" * 40)


arr2 = np.random.randint(1, 10, size=(4, 4))
col_to_del = np.array([1,2,3,4])

print(f'Orig Arr: \n {arr2}')
print("-" * 40)

col_vector = col_to_del[:, np.newaxis]
result = arr2 - col_vector
print(f'Result array: \n {result}')
print("-" * 40)
```
