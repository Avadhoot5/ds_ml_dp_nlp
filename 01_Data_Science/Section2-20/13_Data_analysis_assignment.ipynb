```python
# ### Assignment 10: Masked Arrays

# 1. Create a masked array of shape (4, 4) with random integers and mask the elements greater than 10. Compute the sum of the unmasked elements.
# 2. Create a masked array of shape (3, 3) with random integers and mask the diagonal elements. Replace the masked elements with the mean of the unmasked elements.

import numpy.ma as ma

np.random.seed(42)

raw_data = np.random.randint(1, 21, size=(4,4))
print(f'Raw array: \n {raw_data}')

print("-" * 40)

masked_array = ma.masked_greater(raw_data, 10)

print(f'Masked array \n {masked_array}')
print("-" * 40)

masked_sum = masked_array.sum()
print(f'Sum of masked array: {masked_sum}')

print("-" * 40)

raw_data_2 = np.random.randint(1, 11, size=(3,3))
print(f'Raw array 2: \n {raw_data_2}')
print("-" * 40)

# create a blank mask (all false) and mark diagonal indices as True

mask = np.zeros_like(raw_data_2, dtype=bool)
mask[np.diag_indices_from(mask)] = True

masked_array_2 = ma.masked_array(raw_data_2, mask=mask)
unmasked_mean = masked_array_2.mean()
print(f'Unmasked array \n {masked_array_2}')
print("-" * 40)

final_arr = masked_array_2.filled(fill_value=unmasked_mean)
print("-" * 40)

print(final_arr)
```

```python
# ### Assignment 6: Time Series Analysis

# 1. Create a Pandas DataFrame with a datetime index and one column filled with random integers. Resample the DataFrame to compute the monthly mean of the values.

# 1. Create a datetime index (e.g., daily frequency for a half/full year)
date_index = pd.date_range(start='2026-01-01', end = '2026-06-30', freq='D')

# 2. Create the DataFrame with one column of random integers
df = pd.DataFrame({
    'values': np.random.randint(1, 100, size=len(date_index))
}, index=date_index)

print("Original DataFrame (First 5 rows):")
print(df.head())
print('-' * 50)

# 3. Resample to compute the monthly mean  Note: 'ME' stands for Month End frequency in modern Pandas versions (use 'M' if using older versions)
monthly_mean = df.resample('ME').mean()

print("Monthly Mean DataFrame:")
print(monthly_mean)
print('-' * 50)

# 2. Create a Pandas DataFrame with a datetime index ranging from '2021-01-01' to '2021-12-31' and one column filled with random integers. Compute the rolling mean with a window of 7 days.

date_index_2 = pd.date_range(start='2021-01-01', end = '2021-12-31', freq='D')

df2 = pd.DataFrame({
    'values': np.random.randint(1, 100, size=len(date_index_2))
}, index=date_index_2)

print("Original DataFrame (First 5 rows):")
print(df2.head())
print('-' * 50)

df2['rolling_mean'] = df2['values'].rolling(window=7).mean()

print(df2.head(10))
```

```python
# ### Assignment 10: Working with Text Data

import random
import string

# 1. Create a Pandas Series with 5 random text strings. Convert all the strings to uppercase.

# Helper function to generate 5 random text strings
def get_random_strings():
    return [''.join(random.choices(string.ascii_lowercase, k=6)) for _ in range(5)]


# --- Task 1: Convert all strings to uppercase ---
print("--- Task 1: Uppercase Conversion ---")
series_1 = pd.Series(get_random_strings())

print("Original Series:")
print(series_1)

# Convert to uppercase using .str.upper()
uppercase_series = series_1.str.upper()
print("\nUppercase Series:")
print(uppercase_series)

print('-' * 50)

# 2. Create a Pandas Series with 5 random text strings. Extract the first three characters of each string.

# --- Task 2: Extract the first three characters ---
print("--- Task 2: Extracting Substrings ---")
series_2 = pd.Series(get_random_strings())

print("Original Series:")
print(series_2)

# Extract first three characters using string slicing via .str
extracted_series = series_2.str[:3]
print("\nFirst 3 Characters Extracted:")
print(extracted_series)
```
