```python
# 2. Modify the function to use different logging levels for the file and console handlers.

def basic_logger():
    logger = logging.getLogger('Assignment 2')
    logger.setLevel(logging.DEBUG)
    logger.handlers.clear()
    logger.propagate = False

    formatter = logging.Formatter(
        fmt='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
        datefmt='%Y-%m-%d %H:%M:%S'
    )

    # File handler: logs DEBUG and above
    file_handler = logging.FileHandler(
        '15_Logging_contents/app.log',
        mode='a'
    )
    file_handler.setLevel(logging.DEBUG)
    file_handler.setFormatter(formatter)

    # Console handler: logs WARNING and above
    console_handler = logging.StreamHandler()
    console_handler.setLevel(logging.WARNING)
    console_handler.setFormatter(formatter)

    # Attach handlers to the logger
    logger.addHandler(file_handler)
    logger.addHandler(console_handler)

    # Test all five logging levels
    logger.debug('This is a debug message')
    logger.info('This is an info message')
    logger.warning('This is a warning message')
    logger.error('This is an error message')
    logger.critical('This is a critical message')

    # Close handlers to release resources
    for handler in logger.handlers[:]:
        handler.close()
        logger.removeHandler(handler)


basic_logger()
```

### Assignment 4: Rotating Log Files

```python
# 1. Write a Python function to create a logger that uses a rotating file handler, which creates a new log file when the current log file reaches a certain size.

# 2. Modify the function to keep a specified number of backup log files.

import logging
from logging.handlers import RotatingFileHandler
from pathlib import Path

def create_rotating_logger(
    name: str,
    log_file: str = "15_Logging_contents/app.log",
    max_bytes: int = 5 * 1024 * 1024,  # 5 MB
    backup_count: int = 3
) -> logging.Logger:
    """
    Create a logger that rotates its log file when it reaches a
    specified size and retains a specified number of backup files.

    Args:
        name: Name of the logger.
        log_file: Path to the log file.
        max_bytes: Maximum size of the log file in bytes.
        backup_count: Number of backup log files to retain.

    Returns:
        A configured logging.Logger instance.
    """

    # Ensure the parent directory exists
    Path(log_file).parent.mkdir(parents=True, exist_ok=True)

    # Create or retrieve the logger
    logger = logging.getLogger(name)
    logger.setLevel(logging.DEBUG)

    # Prevent messages from appearing through the root logger
    logger.propagate = False

    # Avoid adding duplicate handlers
    if not any(
        isinstance(handler, RotatingFileHandler)
        for handler in logger.handlers
    ):
        # Create the rotating file handler
        handler = RotatingFileHandler(
            filename=log_file,
            maxBytes=max_bytes,
            backupCount=backup_count,
            encoding="utf-8"
        )

        # Define the log message format
        formatter = logging.Formatter(
            fmt="%(asctime)s - %(name)s - %(levelname)s - %(message)s",
            datefmt="%Y-%m-%d %H:%M:%S"
        )

        handler.setFormatter(formatter)
        handler.setLevel(logging.DEBUG)

        # Attach the handler to the logger
        logger.addHandler(handler)

    return logger


# Create the logger
logger = create_rotating_logger(
    name="MyRotatingLogger",
    log_file="15_Logging_contents/app.log",
    max_bytes=1024,  # 1 KB for testing
    backup_count=2
)

# Generate sample log messages
for i in range(1, 100):
    logger.debug(f"This is log message number {i}")

print("Logging complete. Check the log file and its backups.")
```

### Assignment 6: Contextual Logging

```python
# 1. Write a Python function to create a logger that includes contextual information (e.g., function name, line number) in the log messages.

import logging

def setup_context_logger(logger_name='context_logger'):
    """Creates and configures a logger with contextual info (filename, function name, line number)."""
    logger = logging.getLogger(logger_name)
    logger.setLevel(logging.DEBUG)
    logger.propagate = False

    # Prevent adding duplicate handlers if the function is called multiple times
    if not logger.handlers:
        # Create a console handler
        handler = logging.StreamHandler()
        handler.setLevel(logging.DEBUG)

        # Define the format including contextual placeholders
        formatter = logging.Formatter(
            '%(asctime)s | %(levelname)s | %(filename)s -> %(funcName)s() [Line:%(lineno)d] | %(message)s'
        )
        handler.setFormatter(formatter)
        
        # Attach the handler to the logger
        logger.addHandler(handler)

    return logger

logger = setup_context_logger()

def process_payment(amount):
    logger.info(f"Processing payment of ${amount}")
    if (amount <= 0):
        logger.warning('Amount is invalid')
    else:
        logger.debug('Processing completed')

process_payment(100.0)
process_payment(-5.0)
```

```python
# 2. Modify the function to include additional contextual information (e.g., user ID, session ID).

import logging

def setup_context_logger(logger_name='context_logger'):
    logger = logging.getLogger(logger_name)
    logger.setLevel(logging.DEBUG)
    logger.propagate = False

    if not logger.handlers:
        handler = logging.StreamHandler()
        handler.setLevel(logging.DEBUG)

        formatter = logging.Formatter(
            '%(asctime)s | %(levelname)s | '
            'User: %(user_id)s | Session: %(session_id)s | '
            '%(filename)s -> %(funcName)s() [Line: %(lineno)d] | '
            '%(message)s'
        )

        handler.setFormatter(formatter)
        logger.addHandler(handler)

    return logger


# Create the base logger
logger = setup_context_logger()


def process_payment(amount, user_id, session_id):
    # Add contextual information to the logger
    contextual_logger = logging.LoggerAdapter(
        logger,
        {
            'user_id': user_id,
            'session_id': session_id
        }
    )

    contextual_logger.info(f"Processing payment of ${amount}")

    if amount <= 0:
        contextual_logger.warning("Amount is invalid")
    else:
        contextual_logger.debug("Processing completed")


# Test the logger
process_payment(100.0, user_id='USER101', session_id='SESSION_A1')
process_payment(-5.0, user_id='USER202', session_id='SESSION_B2')
```

```python
# 1. Write a Python function to configure logging using a dictionary. The configuration should include handlers for both file and console logging.

import logging
import logging.config

def configure_logging():
    logging_config = {
        "version": 1,
        "disable_existing_loggers": False,

        "formatters": {
            "standard": {
                "format": "%(asctime)s - %(name)s - %(levelname)s - %(message)s"
            }
        },

        "handlers": {
            "file_handler": {
                "class": "logging.FileHandler",
                "filename": "15_Logging_contents/app.log",
                "level": "DEBUG",
                "formatter": "standard",
            },
            "console_handler": {
                "class": "logging.StreamHandler",
                "level": "INFO",
                "formatter": "standard",
            },
        },

        "root": {
            "level": "DEBUG",
            "handlers": ["file_handler", "console_handler"],
        },
    }

    logging.config.dictConfig(logging_config)


configure_logging()

logger = logging.getLogger(None)

logger.debug("This is a debug message")
logger.info("This is an info message")
logger.warning("This is a warning message")
logger.error("This is an error message")
logger.critical("This is a critical message")
```
