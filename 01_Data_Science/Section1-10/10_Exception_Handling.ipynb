```python
### Assignment 1: Handling Division by Zero
# Write a function that takes two integers as input and returns their division. Use try, except, and finally blocks to handle division by zero and print an appropriate message.

def assig_1(a, b):
    try:
        ans = a / b
        return ans
    except ZeroDivisionError as e:
        print("Error: Cannot divide by zero.")
    finally:
        print("Code executed.")

assig_1(5, 0)
assig_1(10, 5)
```

```python
### Assignment 2: File Reading with Exception Handling
# Write a function that reads the contents of a file named `data.txt`. Use try, except, and finally blocks to handle file not found errors and ensure the file is properly closed.

def assig_2():
    try:
        with open('09_contents/data.txt', 'r') as file:
            contents = file.read()
            return contents
    except FileNotFoundError as e:
        print('File Not found in the directory')
    finally:
        print('Operation Done')

assig_2()
```

```python
### Assignment 3: Handling Multiple Exceptions
# Write a function that takes a list of integers and returns their sum. Use try, except, and finally blocks to handle TypeError if a non-integer value is encountered and print an appropriate message.

def assig_3(lst_int):
    try:
        return sum(lst_int)
    except TypeError as e:
        print('Please provide all integer values in the list format')
    finally:
        print('Function execution done')

print(assig_3([i for i in range(1, 11)]))
print(assig_3([1,2,3,'42', 'test', 5]))
```

```python
### Assignment 4: Exception Handling in User Input
# Write a function that prompts the user to enter an integer. Use try, except, and finally blocks to handle ValueError if the user enters a non-integer value and print an appropriate message.

def assig_4(value):
    try:
        return f'The int value entered is {int(value)}'
    except ValueError:
        print('Please enter an integer value')
    finally:
        print('Function execution done')

print(assig_4(input('Enter a int value:')))
```

```python
### Assignment 5: Exception Handling in Dictionary Access
# Write a function that takes a dictionary and a key as input and returns the value associated with the key. Use try, except, and finally blocks to handle KeyError if the key is not found in the dictionary and print an appropriate message.

def assig_5(ip_dict, ip_key):
    try:
        return f'The value for key {ip_key} is: {ip_dict[ip_key]}'
    except KeyError:
        print('Key not found in the dict')
    finally:
        print('Function execution done')


dict_a = {"Name": "Suresh", "Age": "25", "Title": "analyst", "company": "apple"}

print(assig_5(dict_a, 'Name'))
print(assig_5(dict_a, 'Age'))
print(assig_5(dict_a, 'Gender'))
```

```python
### Assignment 6: Nested Exception Handling
# Write a function that performs nested exception handling. It should first attempt to convert a string to an integer, and then attempt to divide by that integer. Use nested try, except, and finally blocks to handle ValueError and ZeroDivisionError and print appropriate messages.

def assig_6(numerator, string_denominator):
    try:
        convert_value = int(string_denominator)
        try:
            ans = numerator/convert_value
            return ans
        except ZeroDivisionError:
            print('Cannot divide by zero, please provide valid integer value')
        finally:
            print('Divide operation done')
    except ValueError:
        print('Enter an integer value')
    finally:
        print('Function execution done')

print(assig_6(5,'abc'))
print(assig_6(5,'0'))
print(assig_6(5,'5'))
```

```python
### Assignment 7: Exception Handling in List Operations
# Write a function that takes a list and an index as input and returns the element at the given index. Use try, except, and finally blocks to handle IndexError if the index is out of range and print an appropriate message.

def assig_7(ip_lst, ip_idx):
    try:
        return ip_lst[ip_idx]
    except IndexError:
        print('Index out of range, enter valid index')
    finally:
        print('Function execution done')

print(assig_7([1,2,3,'42', 'test', 5], 4))
print(assig_7([1,2,3,'42', 'test', 5], 23))
```

```python
### Assignment 8: Exception Handling in Network Operations
# Write a function that attempts to open a URL and read its contents. Use try, except, and finally blocks to handle network-related errors and print an appropriate message.

import requests

def assig_8(url):
    try:
        response = requests.get(url)
        response.raise_for_status()
        return response.content
    except requests.exceptions.RequestException:
        print('Not able to fetch the requested URL')
    finally:
        print('Function execution done')

print(assig_8('https://www.google.com'))
```

```python
### Assignment 9: Exception Handling in JSON Parsing
# Write a function that attempts to parse a JSON string. Use try, except, and finally blocks to handle JSONDecodeError if the string is not a valid JSON and print an appropriate message.

import json 

def assig_9(data):
    try:
        result = json.loads(data)
        return result
    except json.JSONDecodeError as e:
        print(f'Failed to parse JSON string {e}')
    finally:
        print('Function execution done')

print(assig_9('{"Name": "Suresh", "Age": "25"}'))
print(assig_9('{"Name": "Suresh", "Age": "25"'))
```

```python
### Assignment 10: Custom Exception Handling
# Define a custom exception named `NegativeNumberError`. Write a function that raises this exception if a negative number is encountered in a list. Use try, except, and finally blocks to handle the custom exception and print an appropriate message.

class NegativeNumberError(Exception):
    def __init__(self, message='Negative number encountered in list'):
        self.message = message
        super().__init__(self.message)

def assig_10(ip_lst):
    try:
        for num in ip_lst:
            if num < 0:
                raise NegativeNumberError(f'Negative number encountered in list: {ip_lst}')
        return f'Sucess! Here is the list: {ip_lst}'
    except NegativeNumberError as e:
        return f'Caught Exception -> {e}'
    finally:
        print('Function execution done')

print(assig_10([i for i in range(1, 11)]))
print(assig_10([5,1,-43,-2,4]))
```

```python
### Assignment 11: Exception Handling in Function Calls
# Write a function that calls another function which may raise an exception. Use try, except, and finally blocks to handle the exception and print an appropriate message.

def greet(mssg):
    if (mssg.lower() not in ['morning', 'evening', 'night']):
        raise ValueError('Please add a proper greet message')
    return f'Good {mssg}'

def assig_11(mssg):
    try:
        result = greet(mssg)
        return result
    except ValueError as e:
        print(e)
    finally:
        print('Function execution done')

print(assig_11('Morning'))
print(assig_11('hello'))
```

```python
### Assignment 12: Exception Handling in Class Methods
# Define a class with a method that performs a division operation. Use try, except, and finally blocks within the method to handle division by zero and print an appropriate message.

class Divide:
    def __init__(self, num, deno):
        self.num = num
        self.deno = deno

    def division(self):
        try:
            return f'The result of division is: {self.num/self.deno}'
        except ZeroDivisionError:
            print('Cannot divide by zero, enter an valid integer value')
        finally:
            print('Function execution done')

a = Divide(10, 5)
print(a.division())

b = Divide(10, 0)
print(b.division())
```

```python
### Assignment 13: Exception Handling in Data Conversion
# Write a function that takes a list of strings and converts them to integers. Use try, except, and finally blocks to handle ValueError if a string cannot be converted and print an appropriate message.

def assig_13(ip_lst):
    try:
        convert_lst = list(map(int, ip_lst))
        return convert_lst
    except ValueError:
        print('Cannot convert non-integer strings to int, provide valid integer strings')
    finally:
        print('Function execution done')

print(assig_13([str(i) for i in range(1, 15)]))
print(assig_13([1,2,5,'2', 'test', 8]))
```

```python
### Assignment 14: Exception Handling in List Comprehensions
# Write a function that uses a list comprehension to convert a list of strings to integers. Use try, except, and finally blocks within the list comprehension to handle ValueError and print an appropriate message.

def assig_14(ip_lst):
    try:
        convert_list = [int(i) for i in ip_lst]
        return convert_list
    except ValueError:
        print('Cannot convert non-integer strings to int, provide valid integer strings')
    finally:
        print('Function execution done')

print(assig_14([str(i) for i in range(1, 15)]))
print(assig_14([1,2,5,'2', 'test', 8]))
```

```python
### Assignment 15: Exception Handling in File Writing
# Write a function that attempts to write a list of strings to a file. Use try, except, and finally blocks to handle IOError and ensure the file is properly closed.

def assig_15(ip_lst):
    try:
        with open('09_contents/assig_15.txt', 'w') as file:
            for i in ip_lst:
                file.write(str(i) + ' ')
        print('Done writing to file')
    except IOError:
        print('cannot write to file')
    finally:
        print('Function execution done')

print(assig_15([1,2,5,'2', 'test', 8]))
```
