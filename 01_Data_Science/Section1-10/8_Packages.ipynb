```python
### Assignment 8: Working with the `math` Module
# Use the `math` module to calculate the greatest common divisor (GCD) of two numbers and the factorial of a number.

from math import gcd, factorial

print(gcd(129, 252))
print(factorial(5))
```

```python
### Assignment 9: Working with the `datetime` Module
# Use the `datetime` module to print the current date, calculate the date 100 days from today, and determine the day of the week for a given date.

from datetime import timedelta, datetime as dt 

date_today = dt.date(dt.now())

print(date_today)

print('100days from today')
diff = date_today + timedelta(days=100)
print(diff)

user_input = input('Enter the date in the following format: YYYY-MM-DD')
given_date = dt.fromisoformat(user_input)

day_of_week = given_date.strftime('%A')
print(day_of_week)
```

```python
### Assignment 10: Working with the `random` Module
# Use the `random` module to generate a list of 5 random numbers between 1 and 50 and shuffle the elements of a list.

from random import shuffle, randint

assig_10 = [randint(1, 51) for _ in range(5)]

print(assig_10)
shuffle(assig_10)
print(assig_10)
```
