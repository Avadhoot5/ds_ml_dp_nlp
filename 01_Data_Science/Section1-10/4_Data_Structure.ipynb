```python
### Assignment 9: Matrix Transposition
# Write a function that takes a 3x3 matrix (nested list) as input and returns its transpose. Print the original and transposed matrices.

def transpose_matrix(matrix):
    n = len(matrix)
    new_matrix = [[matrix[row][col] for row in range(n)] for col in range(n)]

    return new_matrix

matrix = [[random.randint(1,5) for _ in range(3)] for _ in range(3)]

print('Input Matrix')
for row in matrix:
    print(row)

print('Output Matrix')
output_matrix = transpose_matrix(matrix)
for row in output_matrix:
    print(row)
```

```python
### Assignment 10: Flattening a Nested List
# Write a function that takes a nested list and flattens it into a single list. Print the original and flattened lists.

nested_list = [[random.randint(1,5) for _ in range(3)] for _ in range(3)]
n = len(nested_list)

flatten_list = [nested_list[row][col] for row in range(n) for col in range(n)]

print('Nested List\n', nested_list)

print('Flattened List\n', flatten_list)
```

```python
### Assignment 14: List Rotation
# Write a function that rotates a list by n positions. Print the original and rotated lists.

new_list = [i for i in range(1,11)]

def reverse_list(arr):
    i, j = 0, len(arr) - 1
    while (i < j):
        arr[i], arr[j] = arr[j], arr[i]
        i += 1
        j -= 1
    return arr

def rotate_by_n(list, d):
    n = len(list)
    d = d % n
    rev_a = reverse_list(list[:d])
    rev_b = reverse_list(list[d:])
    rev_c = rev_a + rev_b
    final_arr = reverse_list(rev_c)
    return final_arr

rotate_list = rotate_by_n(new_list, 3)
print(new_list)
print(rotate_list)
```

```python
### Assignment 15: List Intersection
# Write a function that takes two lists and returns a new list containing only the elements that are present in both lists. Print the intersected list.

list_a = [random.randint(1, 10) for i in range(1,11)]
list_b = [random.randint(1, 10) for i in range(1,11)]

def intersect(a, b):
    return list({element_a for element_a in a if element_a in b})

print(f'List A: {list_a}')
print(f'List b: {list_b}')

print(intersect(list_a, list_b))
```

```python
### Assignment 3: Set Operations
#Create two sets: one with the first 5 positive integers and another with the first 5 even integers. Perform and print the results of union, intersection, difference, and symmetric difference operations on these sets.

positive_integers = {i for i in range(1, 6)}
even_integers = {i for i in range(1, 11) if i % 2 == 0}

print('Pos Integers', positive_integers)
print('Even Integers', even_integers)

# Union

union_set = positive_integers.union(even_integers)
print(union_set)

# Intersection
intersection_set = positive_integers.intersection(even_integers)
print(intersection_set)

# Difference
difference_set = positive_integers.difference(even_integers)
print(difference_set)

# Symmetric Difference
symm_diff = positive_integers.symmetric_difference(even_integers)
print(symm_diff)
```
