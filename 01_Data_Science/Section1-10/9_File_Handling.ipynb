```python
### Assignment 9: Merging Multiple Files
# Write a function that merges the contents of multiple files into a single file named `merged.txt`.

ip_file_list = ['data.txt', 'destination.txt', 'document.txt']

def merge_files(file_list):
    merge_result = ''

    for file_name in file_list:
        with open(f'09_contents/{file_name}', 'r') as file:
            file_result = file.read()
            merge_result += f'''
Contents of file {file_name}:
{file_result}
'''
    with open('09_contents/merged.txt', 'w') as file_write:
        file_write.write(merge_result)

merge_files(ip_file_list)
```

```python
### Assignment 10: Splitting a Large File
# Write a function that splits a large file named `large.txt` into smaller files of 100 lines each.

def split_file(path, lines_per_file=100):
    with open(path, 'r') as file:
        lines = file.readlines()

    for i in range(0, len(lines), lines_per_file):
        chunk = lines[i:i + lines_per_file]
        file_number = (i // lines_per_file) + 1

        with open(f'09_contents/large_part_{file_number}.txt', 'w') as output:
            output.writelines(chunk)

split_file('09_contents/large.txt')
```

```python
### Assignment 11: Creating a Log File
# Write a function that creates a log file named `activity.log` and writes log messages with timestamps.

from datetime import datetime

def log_file():
    with open('09_contents/activity.log', 'a') as file:
        now = datetime.now()
        timestamp = now.strftime("%Y-%m-%d %H:%M:%S")
        file.write(f"{timestamp} - Application started successfully.\n")
        file.write(f"{timestamp} - Low disk space warning.\n")

log_file()
```

```python
### Assignment 13: CSV File Operations
# Write a function that reads a CSV file named `data.csv` and prints its contents as a list of dictionaries.

import csv

def read_csv_content(input_file):
    with open(input_file, mode='r', newline='') as file:
        reader = csv.DictReader(file, delimiter='\t')
        ans = list(reader)
    print(ans)

read_csv_content('09_contents/data.csv')
```

```python
### Assignment 15: File Permission Handling
# Write a function that attempts to read a file named `protected.txt` and handles any permission errors gracefully by printing an error message.

def read_permission(file_name):
    try:
        with open(f'09_contents/{file_name}', 'r') as file:
            contents = file.read()
            print(contents)

    except PermissionError:
        print('Error: You do not have permission to access this file.')

    except FileNotFoundError:
        print('Error: File not found.')


read_permission('protected.txt')
```
